```python
mpg = pd.read_csv("data/mpg.csv")

mpg.groupby(["manufacturer", "drv"])\
.agg(mean_cty = ("cty", "mean"))
```

```python
mpg["drv"].value_counts()
```

```python
mpg.query('category == "suv"') \
.assign(total = mpg["hwy"] + mpg["cty"]) \
.groupby("manufacturer") \
.agg(mean_tot = ("total", "mean")) \
.sort_values("mean_tot", ascending = False).head()
```

```python
mpg.groupby("category") \
.agg(mean_cty = ("cty", "mean")) \
.sort_values("mean_cty", ascending= False)
```

```python
mpg.groupby("category") \
.agg(mean_hwy = ("hwy", "mean")) \
.sort_values("mean_hwy", ascending= False).head(3)
```

```python
mpg.query('category == "compact"') \
    .groupby("manufacturer") \
.agg(n = ) \
.sort_values("mean_hwy", ascending= False).head(3)
```

```python
import pandas as pd
import numpy as np

midwest = pd.read_csv("data/midwest.csv")

midwest["percentage"] = (midwest["poptotal"] - midwest["popadults"]) / midwest["poptotal"] * 100
midwest
```

```python
top5 = midwest.sort_values('percentage', ascending = False).head(5)
top5[['county', 'percentage']]
```

```python
midwest['grade'] = pd.cut(
    midwest['percentage'],
    bins = [0, 10, 20, 30, 100],
    labels = ['1등급', '2등급', '3등급', '4등급']
)
midwest['grade'].value_counts().sort_index()
```

```python
midwest['percentage'] = (
    midwest['popasian'] / midwest['poptotal'] * 100
)

bottom10 = midwest.sort_values(
    'percentage',
    ascending=True
).head(10)
bottom10[['state', 'county', 'percentage']]
```
